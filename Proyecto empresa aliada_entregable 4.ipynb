```python
#CONSULTA PARA VERIFICAR LAS VENTAS UNIENDO VARIAS TABLAS
cur.execute("""
SELECT D.CATEGORY AS CATEGORIA,  F.REGION AS REGION,  DI.YEAR  || '-' || DI.MONTH AS PERIODO, SUM(F.TOTAL_UNIT_SALES) AS TotalVentas
                                FROM DIM_PRODUCT D
                                INNER JOIN FACT_SALES F ON D.ITEM = F.ITEM_CODE
                                INNER JOIN DIM_CALENDAR DI ON F.WEEK = DI.WEEK
                                GROUP BY CATEGORIA, REGION, PERIODO
                                ORDER BY TotalVentas DESC """)
result= cur.fetchall()
result
```

```python
cur.execute("""
SELECT DISTINCT D.CATEGORY ,  F.REGION , SE.SEGMENT
                                FROM DIM_PRODUCT D
                                INNER JOIN FACT_SALES F ON D.ITEM = F.ITEM_CODE
                                INNER JOIN DIM_CALENDAR DI ON F.WEEK = DI.WEEK
                                INNER JOIN DIM_SEGMENT SE ON SE.CATEGORY= D.CATEGORY AND SE.FORMAT=D.FORMAT AND SE.ATTR1=D.ATTR1
                                AND SE.ATTR2=D.ATTR2 AND SE.ATTR3=D.ATTR3
                                ORDER BY F.REGION DESC LIMIT 15 """)
result= cur.fetchall()
result
```
